```python
import matplotlib.pyplot as plt
import numpy as np


class DenseLayerFromScratch:
    def __init__(self, input_dim, units, activation="relu"):
        self.W = np.random.randn(units, input_dim) * 0.01
        self.b = np.zeros((units, 1))
        self.activation = activation

    def _relu(self, Z):
        return np.maximum(0, Z)

    def _sigmoid(self, Z):
        return 1 / (1 + np.exp(-Z))

    def forward(self, A_prev):
        #  Z = W * A_prev + b
        self.Z = np.dot(self.W, A_prev) + self.b

        if self.activation == "relu":
            self.A = self._relu(self.Z)
        elif self.activation == "sigmoid":
            self.A = self._sigmoid(self.Z)
        else:
            self.A = self.Z  # Linear

        return self.A


if __name__ == "__main__":
    np.random.seed(42)
    sample_inputs = np.random.rand(3, 5)  # [Features x Samples]

    layer1 = DenseLayerFromScratch(input_dim=3, units=4, activation="relu")
    output_activations = layer1.forward(sample_inputs)


    print("Sample Inputs")
    print(sample_inputs)
    print("\nOutput Activations")
    print(output_activations)

    plt.figure(figsize=(8, 4))
    plt.imshow(output_activations, cmap="viridis", aspect="auto")
    plt.colorbar(label="Activation Value")
    plt.title("Layer Activation Output Simulation")
    plt.xlabel("Sample Index")
    plt.ylabel("Neuron Unit Index")
```
