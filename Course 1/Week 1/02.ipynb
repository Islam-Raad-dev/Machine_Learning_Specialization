```python
import numpy as np
import matplotlib.pyplot as plt
                              
x = np.linspace(0, 3, 100)

cases = [
    {"w": 0.0, "b": 1.5, "title": r"$f(x) = 0 \cdot x + 1.5$"},
    {"w": 0.5, "b": 0.0, "title": r"$f(x) = 0.5 \cdot x$"},
    {"w": 0.5, "b": 1.0, "title": r"$f(x) = 0.5 \cdot x + 1$"}
]
           
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

for idx, case in enumerate(cases):
    w, b = case["w"], case["b"]
    y = w * x + b
    
    ax = axes[idx]
    ax.plot(x, y, color='#007acc', linewidth=3)
    ax.set_xlim(0, 3)
    ax.set_ylim(0, 3)
    ax.set_xticks([0, 1, 2, 3])
    ax.set_yticks([0, 1, 2, 3])
    ax.set_title(case["title"], fontsize=12)
    ax.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
```
