```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, LogisticRegression



np.random.seed(42)
X_reg = 2 * np.random.rand(100, 1) 
y_reg = 4 + 3 * X_reg + np.random.randn(100, 1) 

reg_model = LinearRegression()
reg_model.fit(X_reg, y_reg)
X_new_reg = np.array([[0], [2]])
y_predict_reg = reg_model.predict(X_new_reg)




X_class = np.linspace(0, 10, 100).reshape(-1, 1) 
y_class = (X_class.ravel() > 5).astype(int)

clf_model = LogisticRegression()
clf_model.fit(X_class, y_class)
y_prob = clf_model.predict_proba(X_class)[:, 1]



fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].scatter(X_reg, y_reg, color='blue', alpha=0.6, label='Data Points (Homes)')
axes[0].plot(X_new_reg, y_predict_reg, color='red', linewidth=2, label='Linear Fit (Prediction)')
axes[0].set_title('Regression: Predicting Continuous Price')
axes[0].set_xlabel('House Size')
axes[0].set_ylabel('Price')
axes[0].legend()
axes[0].grid(True)

axes[1].scatter(X_class, y_class, color='green', alpha=0.6, label='Actual Data (0: Benign, 1: Malignant)')
axes[1].plot(X_class, y_prob, color='magenta', linewidth=2, label='Sigmoid Probability Curve')
axes[1].axhline(0.5, color='black', linestyle='--', label='Threshold (0.5)')
axes[1].set_title('Classification: Tumor Malignancy Probability')
axes[1].set_xlabel('Tumor Size')
axes[1].set_ylabel('Probability / Class')
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.savefig('/tmp/regression_vs_classification.png')
print("Simulation complete. Image saved to /tmp/regression_vs_classification.png")
```
