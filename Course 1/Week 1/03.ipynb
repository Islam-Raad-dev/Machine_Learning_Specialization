```python
import matplotlib.pyplot as plt
import numpy as np
       
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 3.5, 5, 6.5, 8])

w = 1.5
b_values = [0, 2, 5, -2] 

x_line = np.linspace(0, 6, 100)

plt.figure(figsize=(8, 5))
plt.scatter(x, y, color="red", label="Data Points (y)", zorder=5)

for b in b_values:
    y_line = w * x_line + b
    plt.plot(x_line, y_line, label=f"Line: f(x) = {w}x + {b}")

plt.title("Effect of changing Bias (b) on Linear Model")
plt.xlabel("Input (x)")
plt.ylabel("Prediction (y)")
plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)
plt.grid(True)
plt.legend()
```
